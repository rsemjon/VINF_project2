# Books with both a crawled and a wiki author

The joined CSV has no plain `author` column, so this compares `crawled_author` with `wiki_author`.

```python
import pandas as pd

pd.set_option('display.max_colwidth', 80)

# same options as the write in extractor_wiki.py, empty cells stay ''
df = pd.read_csv(
    'data/wiki_join_craw_attempt1.csv',
    sep='	', quotechar='"', doublequote=True,
    dtype=str, keep_default_na=False,
)
df.shape
```

```python
has_crawled = df['crawled_author'].str.strip() != ''
has_wiki = df['wiki_author'].str.strip() != ''

pd.Series({
    'rows': len(df),
    'crawled_author filled': has_crawled.sum(),
    'wiki_author filled': has_wiki.sum(),
    'wiki pages (page_id filled)': (df['page_id'] != '').sum(),
    'both authors filled': (has_crawled & has_wiki).sum(),
})
```

```python
both = df[has_crawled & has_wiki]
both[['title', 'crawled_author', 'wiki_author']]
```
